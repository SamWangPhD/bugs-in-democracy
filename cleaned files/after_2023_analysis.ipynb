```python
import seaborn as sns
import os
from scipy.stats import kurtosis, skew

from rcv_distribution import *
from MDS_analysis import *
from voting_rules import *
from consistency import *
```

```python
directory = "dataverse_files_only_after_2023"
election_table = "election_table_090924.csv"
election_df = pd.read_csv(election_table,
                          encoding="cp1252",  # or "latin1"
                          engine="python")   
gamma_values = pd.read_csv("null_gamma_values3.csv")
```

```python
for filename in os.listdir(directory):
    try:
        ballots, candidates = parse_election_data(("dataverse_files_only_after_2023"+ "/" + filename))
        print("candidates: ", candidates)
        if len(candidates) > 1:
            test = perform_rcv_analysis(ballots, candidates, n_init=100, max_itr=1000, n_runs=1000, metric=False)
            mds_1d_coordinates, mds_2d_coordinates, most_common_order, order_frequencies, candidate_names = test
            
            # Print the normalized distances between candidates and plot the MDS analysis
            normalized_distances = get_distances_normalized(most_common_order, mds_1d_coordinates, candidate_names)

            #print("new distances: ", normalized_distances_original)
            consistent_ballots, gamma_permissive = get_permissive_gamma(ballots, normalized_distances)
            gamma_strict = get_strict_gamma(ballots, normalized_distances)

            new_row = {
                "filename": filename,
                "gamma": gamma_permissive,
                "candidates": len(candidates)
            }

            new_gamma_row = {
                "filname": filename,
                "#candidates": len(candidates),
                "permissive_gamma": gamma_permissive,
                "strict_gamma": gamma_strict
            }
        else:
            new_row = {
                "filename": filename,
                "gamma": 1.0,
                "candidates": len(candidates)
            }

            new_gamma_row = {
                "filename": filename,
                "#candidates": len(candidates),
                "permissive_gamma": 1.0,
                "strict_gamma": 1.0
            }
        
        election_df.loc[len(election_df)] = new_row
        gamma_values.loc[len(gamma_values)] = new_gamma_row
        print(filename, ": ", gamma_permissive, len(candidates))
    except Exception as e:
        # Print the summary of the error
        print(filename, " ", e)

    






```

```python
election_df.to_csv("election_table_after_2023.csv", index=False)
```

```python
gamma_values.to_csv("gamma_values_after_2023.csv", index=False)
```
